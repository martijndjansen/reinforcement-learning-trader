```python
import yfinance as yf
import pandas as pd
dat = pd.DataFrame(yf.download(["ASML","NVDA","SPY"], period='1mo',auto_adjust=True))
```

```python
stocks: list[str] = dat["Close"].columns.tolist() # type: ignore
import numpy as np
for stock in stocks:
    dat["Return", stock] = np.log(dat["Close", stock] / dat["Close", stock].shift(1)).fillna(0) # type: ignore
```
